# Visualization on Iris Datasets

### loading the libraries

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

df = pd.read_csv("Iris.csv")
print(df.head())
```

## Univariate Analysis

### 1. Histogram

```python
plt.figure(figsize=(8, 5))
sns.histplot(data=df, x="SepalLengthCm", kde=True, bins=15)
plt.title("Distribution of Sepal Length")
plt.xlabel("Sepal Length (cm)")
plt.ylabel("Count")
plt.show()
```

### Box Plot for Outliers

```python
plt.figure(figsize=(10, 6))
sns.boxplot(data=df.drop(columns=["Id", "Species"]))
plt.title("Box Plot of Iris Physical Features")
plt.ylabel("Measurement (cm)")
plt.show()
```

### Bar Graph

```python
df.Species.value_counts(normalize=True)
df.Species.value_counts(normalize=True).plot.barh()
plt.title("Percentage Distribution of Iris Species")
plt.xlabel("Proportion")
plt.ylabel("Species")
plt.show()
```

### Pie Chart

```python
# Calculate the percentage of each species category
df.Species.value_counts(normalize=True)

# Plot the pie chart of species categories
df.Species.value_counts(normalize=True).plot.pie(autopct="%1.1f%%")
plt.title("Proportion of Species in Dataset")
plt.ylabel("Species")
plt.show()
```

## Bivariate Analysis

### Numeric-Numeric Scatter Plots

```python
#Scatter plot 1: Sepal Length vs Sepal Width
plt.figure(figsize=(8, 5))
plt.scatter(df["SepalLengthCm"], df["SepalWidthCm"])
plt.title("Scatter Plot of Sepal Length vs Sepal Width")
plt.xlabel("SepalLengthCm")
plt.ylabel("SepalWidthCm")
plt.show()

# Scatter Plot 2: Petal Length vs Petal Width
plt.figure(figsize=(8, 5))
plt.scatter(df["PetalLengthCm"], df["PetalWidthCm"])
plt.title("Scatter Plot of Petal Length vs Petal Width")
plt.xlabel("PetalLengthCm")
plt.ylabel("PetalWidthCm")
plt.show()
```

### Pair plot

```python
sns.pairplot(data=df, vars=["SepalLengthCm", "SepalWidthCm", "PetalLengthCm", "PetalWidthCm"])
plt.suptitle("Pairplots of Iris Physical Measurements", y=1.02)
plt.show()
```

### Numerical categorical analysis (Boxplot)

```python
plt.figure(figsize=(8, 6))

#box plot of Species vs Sepal Length
sns.boxplot(data=df, x="Species", y="SepalLengthCm")

plt.title("Boxplot of Sepal Length with respect to Species")
plt.xlabel("Species")
plt.ylabel("SepalLengthCm")
plt.show()
```

```python
import numpy as np 

df["response_rate"] = np.where(df["Species"] == "Iris-setosa", 1, 0)

df["sepal_type"] = np.where(df["SepalWidthCm"] > 3.0, "wide", "narrow")

df.groupby("sepal_type")["response_rate"].mean().plot.bar()
plt.title("Bar Graph of Sepal Type with Average Value of Response Rate")
plt.xlabel("Sepal Type")
plt.ylabel("Proportion of Setosa")
plt.xticks(rotation=0)
plt.show()
```
